```python
%sql
CREATE CATALOG NycTaxiData;
```

```python
%sql
CREATE SCHEMA NycTaxiData.gold;
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
silveContainer = "abfss://silverlayer@nyctaxiprojectdatalake.dfs.core.windows.net/"
goldContainer = "abfss://goldlayer@nyctaxiprojectdatalake.dfs.core.windows.net/"
```

```python
df_tripType = spark.read.format('parquet')\
                        .options(inferSchema = True, header = True)\
                        .load(f'{silveContainer}/trip_type')
```

```python
df_tripType.display()
```

```python
df_tripType.write \
    .format('delta') \
    .mode('append') \
    .saveAsTable('NycTaxiData.gold.trip_type')
```

```python
df_tripZone = spark.read.format('parquet')\
                        .options(inferSchema = True, header = True)\
                        .load(f'{silveContainer}/trip_zone')
```

```python
df_tripZone.display()
```

```python
df_tripZone.write.format('delta')\
                 .saveAsTable('NycTaxiData.gold.trip_zone')
```
